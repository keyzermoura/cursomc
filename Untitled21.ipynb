```python
%%writefile /content/investidor_app/lib/main.dart
import 'package:flutter/material.dart';
import 'package:fl_chart/fl_chart.dart';
import 'dart:convert';
import 'package:http/http.dart' as http;

void main() {
  runApp(const InvestidorApp());
}

class AtivoModel {
  final String ticker;
  final String nome;
  final String classe;
  final String setor;
  double quantidade;
  double precoMedio;
  double precoAtual;
  double variacaoDia;

  AtivoModel({
    required this.ticker,
    required this.nome,
    required this.classe,
    required this.setor,
    required this.quantidade,
    required this.precoMedio,
    this.precoAtual = 0.0,
    this.variacaoDia = 0.0,
  });

  double get valorTotal => quantidade * (precoAtual > 0 ? precoAtual : precoMedio);
  double get totalInvestido => quantidade * precoMedio;
  double get lucroPrejuizo => valorTotal - totalInvestido;
  double get rentabilidadePct => totalInvestido > 0 ? (lucroPrejuizo / totalInvestido) * 100 : 0.0;
}

class InvestidorApp extends StatelessWidget {
  const InvestidorApp({super.key});

  @override
  Widget build(BuildContext context) {
    return MaterialApp(
      debugShowCheckedModeBanner: false,
      title: 'Investidor IA',
      theme: ThemeData.dark().copyWith(
        scaffoldBackgroundColor: const Color(0xFF0D1117),
        cardColor: const Color(0xFF161B22),
        primaryColor: const Color(0xFF00E676),
      ),
      home: const DashboardScreen(),
    );
  }
}

class DashboardScreen extends StatefulWidget {
  const DashboardScreen({super.key});

  @override
  State<DashboardScreen> createState() => _DashboardScreenState();
}

class _DashboardScreenState extends State<DashboardScreen> {
  bool _ocultarValores = false;
  bool _carregandoCotacoes = false;

  final List<AtivoModel> _carteira = [
    AtivoModel(ticker: 'PETR4', nome: 'Petrobras PN', classe: 'Ações', setor: 'Energia', quantidade: 150, precoMedio: 32.50, precoAtual: 38.50, variacaoDia: 1.45),
    AtivoModel(ticker: 'VALE3', nome: 'Vale ON', classe: 'Ações', setor: 'Mineração', quantidade: 80, precoMedio: 68.00, precoAtual: 62.10, variacaoDia: -0.85),
    AtivoModel(ticker: 'MXRF11', nome: 'Maxi Renda FII', classe: 'FIIs', setor: 'Papel', quantidade: 500, precoMedio: 10.10, precoAtual: 10.45, variacaoDia: 0.10),
  ];

  double get patrimonioTotal => _carteira.fold(0, (sum, item) => sum + item.valorTotal);
  double get totalInvestido => _carteira.fold(0, (sum, item) => sum + item.totalInvestido);
  double get lucroTotal => patrimonioTotal - totalInvestido;
  double get rentabilidadePct => totalInvestido > 0 ? (lucroTotal / totalInvestido) * 100 : 0;

  Future<void> _atualizarCotacoes() async {
    setState(() => _carregandoCotacoes = true);
    for (var ativo in _carteira) {
      try {
        final url = Uri.parse('https://brapi.dev/api/quote/${ativo.ticker}');
        final response = await http.get(url);
        if (response.statusCode == 200) {
          final data = json.decode(response.body);
          if (data['results'] != null && (data['results'] as List).isNotEmpty) {
            final result = data['results'][0];
            setState(() {
              ativo.precoAtual = (result['regularMarketPrice'] as num).toDouble();
              ativo.variacaoDia = (result['regularMarketChangePercent'] as num).toDouble();
            });
          }
        }
      } catch (e) {
        print('Erro no ticker ${ativo.ticker}: $e');
      }
    }
    setState(() => _carregandoCotacoes = false);
  }

  @override
  Widget build(BuildContext context) {
    return Scaffold(
      appBar: AppBar(
        backgroundColor: const Color(0xFF161B22),
        elevation: 0,
        title: const Text('Investidor IA', style: TextStyle(fontWeight: FontWeight.bold)),
        actions: [
          IconButton(
            icon: Icon(_ocultarValores ? Icons.visibility_off : Icons.visibility),
            onPressed: () => setState(() => _ocultarValores = !_ocultarValores),
          ),
          IconButton(
            icon: _carregandoCotacoes
                ? const SizedBox(width: 20, height: 20, child: CircularProgressIndicator(strokeWidth: 2))
                : const Icon(Icons.refresh),
            onPressed: _atualizarCotacoes,
          ),
        ],
      ),
      body: ListView(
        padding: const EdgeInsets.all(16),
        children: [
          _buildResumoPatrimonio(),
          const SizedBox(height: 20),
          _buildGraficoAlocacao(),
          const SizedBox(height: 24),
          Row(
            mainAxisAlignment: MainAxisAlignment.spaceBetween,
            children: [
              const Text('Meus Ativos', style: TextStyle(fontSize: 18, fontWeight: FontWeight.bold)),
              ElevatedButton.icon(
                style: ElevatedButton.styleFrom(backgroundColor: const Color(0xFF00E676)),
                onPressed: () {},
                icon: const Icon(Icons.add, color: Colors.black),
                label: const Text('Adicionar', style: TextStyle(color: Colors.black, fontWeight: FontWeight.bold)),
              ),
            ],
          ),
          const SizedBox(height: 12),
          ..._carteira.map((ativo) => _buildCardAtivo(ativo)),
        ],
      ),
    );
  }

  Widget _buildResumoPatrimonio() {
    return Container(
      padding: const EdgeInsets.all(20),
      decoration: BoxDecoration(
        color: const Color(0xFF161B22),
        borderRadius: BorderRadius.circular(16),
        border: Border.all(color: Colors.white10),
      ),
      child: Column(
        crossAxisAlignment: CrossAxisAlignment.start,
        children: [
          const Text('Patrimônio Total', style: TextStyle(color: Colors.white60, fontSize: 14)),
          const SizedBox(height: 8),
          Text(
            _ocultarValores ? 'R\$ ••••••••' : 'R\$ ${patrimonioTotal.toStringAsFixed(2)}',
            style: const TextStyle(fontSize: 30, fontWeight: FontWeight.bold, color: Colors.white),
          ),
          const SizedBox(height: 16),
          Row(
            mainAxisAlignment: MainAxisAlignment.spaceBetween,
            children: [
              Column(
                crossAxisAlignment: CrossAxisAlignment.start,
                children: [
                  const Text('Rentabilidade', style: TextStyle(color: Colors.white54, fontSize: 12)),
                  Text(
                    '${rentabilidadePct >= 0 ? "+" : ""}${rentabilidadePct.toStringAsFixed(2)}%',
                    style: TextStyle(
                      color: rentabilidadePct >= 0 ? const Color(0xFF00E676) : Colors.redAccent,
                      fontWeight: FontWeight.bold,
                      fontSize: 16,
                    ),
                  ),
                ],
              ),
              Column(
                crossAxisAlignment: CrossAxisAlignment.end,
                children: [
                  const Text('Lucro Total', style: TextStyle(color: Colors.white54, fontSize: 12)),
                  Text(
                    _ocultarValores ? 'R\$ •••' : 'R\$ ${lucroTotal.toStringAsFixed(2)}',
                    style: TextStyle(
                      color: lucroTotal >= 0 ? const Color(0xFF00E676) : Colors.redAccent,
                      fontWeight: FontWeight.bold,
                      fontSize: 16,
                    ),
                  ),
                ],
              ),
            ],
          ),
        ],
      ),
    );
  }

  Widget _buildGraficoAlocacao() {
    if (patrimonioTotal == 0) return const SizedBox.shrink();

    return Container(
      height: 200,
      padding: const EdgeInsets.all(16),
      decoration: BoxDecoration(
        color: const Color(0xFF161B22),
        borderRadius: BorderRadius.circular(16),
        border: Border.all(color: Colors.white10),
      ),
      child: Row(
        children: [
          Expanded(
            child: PieChart(
              PieChartData(
                sectionsSpace: 4,
                centerSpaceRadius: 40,
                sections: _carteira.map((ativo) {
                  final pct = (ativo.valorTotal / patrimonioTotal) * 100;
                  return PieChartSectionData(
                    value: ativo.valorTotal,
                    title: '${pct.toStringAsFixed(0)}%',
                    radius: 35,
                    titleStyle: const TextStyle(fontSize: 12, fontWeight: FontWeight.bold, color: Colors.white),
                  );
                }).toList(),
              ),
            ),
          ),
          Column(
            mainAxisAlignment: MainAxisAlignment.center,
            crossAxisAlignment: CrossAxisAlignment.start,
            children: _carteira.map((a) => Padding(
              padding: const EdgeInsets.symmetric(vertical: 4),
              child: Row(
                children: [
                  Text('${a.ticker}: ', style: const TextStyle(fontWeight: FontWeight.bold)),
                  Text('${((a.valorTotal / patrimonioTotal) * 100).toStringAsFixed(1)}%', style: const TextStyle(color: Colors.white70)),
                ],
              ),
            )).toList(),
          )
        ],
      ),
    );
  }

  Widget _buildCardAtivo(AtivoModel ativo) {
    final pesoPct = patrimonioTotal > 0 ? (ativo.valorTotal / patrimonioTotal) * 100 : 0.0;
    final variacaoPositiva = ativo.variacaoDia >= 0;

    return Card(
      margin: const EdgeInsets.only(bottom: 12),
      shape: RoundedRectangleBorder(borderRadius: BorderRadius.circular(12)),
      child: ListTile(
        contentPadding: const EdgeInsets.symmetric(horizontal: 16, vertical: 8),
        leading: CircleAvatar(
          backgroundColor: const Color(0xFF00E676).withOpacity(0.1),
          child: Text(ativo.ticker.substring(0, 2), style: const TextStyle(color: Color(0xFF00E676), fontWeight: FontWeight.bold)),
        ),
        title: Row(
          children: [
            Text(ativo.ticker, style: const TextStyle(fontWeight: FontWeight.bold)),
            const SizedBox(width: 8),
            Container(
              padding: const EdgeInsets.symmetric(horizontal: 6, vertical: 2),
              decoration: BoxDecoration(color: Colors.white10, borderRadius: BorderRadius.circular(4)),
              child: Text(ativo.classe, style: const TextStyle(fontSize: 10, color: Colors.white70)),
            )
          ],
        ),
        subtitle: Text('Peso: ${pesoPct.toStringAsFixed(1)}% | Qtd: ${ativo.quantidade.toInt()}', style: const TextStyle(color: Colors.white54, fontSize: 12)),
        trailing: Column(
          mainAxisAlignment: MainAxisAlignment.center,
          crossAxisAlignment: CrossAxisAlignment.end,
          children: [
            Text(
              _ocultarValores ? 'R\$ •••' : 'R\$ ${ativo.precoAtual.toStringAsFixed(2)}',
              style: const TextStyle(fontWeight: FontWeight.bold, fontSize: 15),
            ),
            Text(
              '${variacaoPositiva ? "+" : ""}${ativo.variacaoDia.toStringAsFixed(2)}%',
              style: TextStyle(
                color: variacaoPositiva ? const Color(0xFF00E676) : Colors.redAccent,
                fontWeight: FontWeight.bold,
                fontSize: 12,
              ),
            ),
          ],
        ),
      ),
    );
  }
}
```

```python
%%writefile /content/investidor_app/lib/main.dart
import 'package:flutter/material.dart';
import 'package:fl_chart/fl_chart.dart';
import 'dart:convert';
import 'package:http/http.dart' as http;
import 'package:shared_preferences/shared_preferences.dart';

void main() {
  runApp(const InvestidorApp());
}

class AtivoModel {
  final String ticker;
  final String nome;
  final String classe;
  final String setor;
  double quantidade;
  double precoMedio;
  double precoAtual;
  double variacaoDia;

  AtivoModel({
    required this.ticker,
    required this.nome,
    required this.classe,
    required this.setor,
    required this.quantidade,
    required this.precoMedio,
    this.precoAtual = 0.0,
    this.variacaoDia = 0.0,
  });

  double get valorTotal => quantidade * (precoAtual > 0 ? precoAtual : precoMedio);
  double get totalInvestido => quantidade * precoMedio;
  double get lucroPrejuizo => valorTotal - totalInvestido;
  double get rentabilidadePct => totalInvestido > 0 ? (lucroPrejuizo / totalInvestido) * 100 : 0.0;

  Map<String, dynamic> toMap() {
    return {
      'ticker': ticker,
      'nome': nome,
      'classe': classe,
      'setor': setor,
      'quantidade': quantidade,
      'precoMedio': precoMedio,
    };
  }

  factory AtivoModel.fromMap(Map<String, dynamic> map) {
    return AtivoModel(
      ticker: map['ticker'] ?? '',
      nome: map['nome'] ?? map['ticker'] ?? '',
      classe: map['classe'] ?? 'Ações',
      setor: map['setor'] ?? 'Geral',
      quantidade: (map['quantidade'] as num).toDouble(),
      precoMedio: (map['precoMedio'] as num).toDouble(),
      precoAtual: (map['precoMedio'] as num).toDouble(),
    );
  }
}

class InvestidorApp extends StatelessWidget {
  const InvestidorApp({super.key});

  @override
  Widget build(BuildContext context) {
    return MaterialApp(
      debugShowCheckedModeBanner: false,
      title: 'Investidor IA',
      theme: ThemeData.dark().copyWith(
        scaffoldBackgroundColor: const Color(0xFF0D1117),
        cardColor: const Color(0xFF161B22),
        primaryColor: const Color(0xFF00E676),
      ),
      home: const DashboardScreen(),
    );
  }
}

class DashboardScreen extends StatefulWidget {
  const DashboardScreen({super.key});

  @override
  State<DashboardScreen> createState() => _DashboardScreenState();
}

class _DashboardScreenState extends State<DashboardScreen> {
  bool _ocultarValores = false;
  bool _carregandoCotacoes = false;
  List<AtivoModel> _carteira = [];

  @override
  void initState() {
    super.initState();
    _carregarCarteiraLocal();
  }

  Future<void> _salvarCarteiraLocal() async {
    final prefs = await SharedPreferences.getInstance();
    final List<String> carteiraJson = _carteira.map((item) => json.encode(item.toMap())).toList();
    await prefs.setStringList('minha_carteira', carteiraJson);
  }

  Future<void> _carregarCarteiraLocal() async {
    final prefs = await SharedPreferences.getInstance();
    final List<String>? carteiraJson = prefs.getStringList('minha_carteira');

    if (carteiraJson != null && carteiraJson.isNotEmpty) {
      setState(() {
        _carteira = carteiraJson.map((str) => AtivoModel.fromMap(json.decode(str))).toList();
      });
    } else {
      setState(() {
        _carteira = [
          AtivoModel(ticker: 'PETR4', nome: 'Petrobras PN', classe: 'Ações', setor: 'Energia', quantidade: 150, precoMedio: 32.50, precoAtual: 38.50, variacaoDia: 1.45),
          AtivoModel(ticker: 'VALE3', nome: 'Vale ON', classe: 'Ações', setor: 'Mineração', quantidade: 80, precoMedio: 68.00, precoAtual: 62.10, variacaoDia: -0.85),
          AtivoModel(ticker: 'MXRF11', nome: 'Maxi Renda FII', classe: 'FIIs', setor: 'Papel', quantidade: 500, precoMedio: 10.10, precoAtual: 10.45, variacaoDia: 0.10),
        ];
      });
      _salvarCarteiraLocal();
    }
    _atualizarCotacoes();
  }

  double get patrimonioTotal => _carteira.fold(0, (sum, item) => sum + item.valorTotal);
  double get totalInvestido => _carteira.fold(0, (sum, item) => sum + item.totalInvestido);
  double get lucroTotal => patrimonioTotal - totalInvestido;
  double get rentabilidadePct => totalInvestido > 0 ? (lucroTotal / totalInvestido) * 100 : 0;

  Future<void> _atualizarCotacoes() async {
    if (_carteira.isEmpty) return;
    setState(() => _carregandoCotacoes = true);

    for (var ativo in _carteira) {
      try {
        final url = Uri.parse('https://brapi.dev/api/quote/${ativo.ticker}');
        final response = await http.get(url);
        if (response.statusCode == 200) {
          final data = json.decode(response.body);
          if (data['results'] != null && (data['results'] as List).isNotEmpty) {
            final result = data['results'][0];
            setState(() {
              ativo.precoAtual = (result['regularMarketPrice'] as num).toDouble();
              ativo.variacaoDia = (result['regularMarketChangePercent'] as num).toDouble();
            });
          }
        }
      } catch (e) {
        print('Erro no ticker ${ativo.ticker}: $e');
      }
    }
    setState(() => _carregandoCotacoes = false);
  }

  void _removerAtivo(int index) {
    setState(() {
      _carteira.removeAt(index);
    });
    _salvarCarteiraLocal();
  }

  void _exibirModalAdicionarAtivo() {
    final tickerCtrl = TextEditingController();
    final qtdCtrl = TextEditingController();
    final precoCtrl = TextEditingController();
    String classeSel = 'Ações';

    showModalBottomSheet(
      context: context,
      isScrollControlled: true,
      backgroundColor: const Color(0xFF161B22),
      shape: const RoundedRectangleBorder(
        borderRadius: BorderRadius.vertical(top: Radius.circular(20)),
      ),
      builder: (ctx) => StatefulBuilder(
        builder: (ctxState, setModalState) => Padding(
          padding: EdgeInsets.only(
            top: 20, left: 20, right: 20,
            bottom: MediaQuery.of(ctx).viewInsets.bottom + 20,
          ),
          child: Column(
            mainAxisSize: MainAxisSize.min,
            crossAxisAlignment: CrossAxisAlignment.start,
            children: [
              const Text('Adicionar Novo Ativo', style: TextStyle(fontSize: 18, fontWeight: FontWeight.bold)),
              const SizedBox(height: 16),
              TextField(
                controller: tickerCtrl,
                textCapitalization: TextCapitalization.characters,
                decoration: const InputDecoration(
                  labelText: 'Ticker (ex: WEGE3, HGLG11)',
                  border: OutlineInputBorder(),
                ),
              ),
              const SizedBox(height: 12),
              DropdownButtonFormField<String>(
                value: classeSel,
                decoration: const InputDecoration(labelText: 'Classe', border: OutlineInputBorder()),
                items: ['Ações', 'FIIs', 'Renda Fixa', 'BDRs'].map((c) => DropdownMenuItem(value: c, child: Text(c))).toList(),
                onChanged: (val) => setModalState(() => classeSel = val!),
              ),
              const SizedBox(height: 12),
              Row(
                children: [
                  Expanded(
                    child: TextField(
                      controller: qtdCtrl,
                      keyboardType: TextInputType.number,
                      decoration: const InputDecoration(labelText: 'Quantidade', border: OutlineInputBorder()),
                    ),
                  ),
                  const SizedBox(width: 12),
                  Expanded(
                    child: TextField(
                      controller: precoCtrl,
                      keyboardType: const TextInputType.numberWithOptions(decimal: true),
                      decoration: const InputDecoration(labelText: 'Preço Médio (R\$)', border: OutlineInputBorder()),
                    ),
                  ),
                ],
              ),
              const SizedBox(height: 20),
              ElevatedButton(
                style: ElevatedButton.styleFrom(
                  backgroundColor: const Color(0xFF00E676),
                  minimumSize: const Size.fromHeight(50),
                ),
                onPressed: () {
                  final ticker = tickerCtrl.text.trim().toUpperCase();
                  final qtd = double.tryParse(qtdCtrl.text) ?? 0;
                  final preco = double.tryParse(precoCtrl.text.replaceAll(',', '.')) ?? 0;

                  if (ticker.isNotEmpty && qtd > 0 && preco > 0) {
                    final novoAtivo = AtivoModel(
                      ticker: ticker,
                      nome: ticker,
                      classe: classeSel,
                      setor: 'Geral',
                      quantidade: qtd,
                      precoMedio: preco,
                      precoAtual: preco,
                    );
                    setState(() {
                      _carteira.add(novoAtivo);
                    });
                    _salvarCarteiraLocal();
                    Navigator.pop(ctx);
                    _atualizarCotacoes();
                  }
                },
                child: const Text('Salvar Ativo', style: TextStyle(color: Colors.black, fontWeight: FontWeight.bold)),
              )
            ],
          ),
        ),
      ),
    );
  }

  @override
  Widget build(BuildContext context) {
    return Scaffold(
      appBar: AppBar(
        backgroundColor: const Color(0xFF161B22),
        elevation: 0,
        title: const Text('Investidor IA', style: TextStyle(fontWeight: FontWeight.bold)),
        actions: [
          IconButton(
            icon: Icon(_ocultarValores ? Icons.visibility_off : Icons.visibility),
            onPressed: () => setState(() => _ocultarValores = !_ocultarValores),
          ),
          IconButton(
            icon: _carregandoCotacoes
                ? const SizedBox(width: 20, height: 20, child: CircularProgressIndicator(strokeWidth: 2))
                : const Icon(Icons.refresh),
            onPressed: _atualizarCotacoes,
          ),
        ],
      ),
      body: ListView(
        padding: const EdgeInsets.all(16),
        children: [
          _buildResumoPatrimonio(),
          const SizedBox(height: 20),
          _buildGraficoAlocacao(),
          const SizedBox(height: 24),
          Row(
            mainAxisAlignment: MainAxisAlignment.spaceBetween,
            children: [
              const Text('Meus Ativos', style: TextStyle(fontSize: 18, fontWeight: FontWeight.bold)),
              ElevatedButton.icon(
                style: ElevatedButton.styleFrom(backgroundColor: const Color(0xFF00E676)),
                onPressed: _exibirModalAdicionarAtivo,
                icon: const Icon(Icons.add, color: Colors.black),
                label: const Text('Adicionar', style: TextStyle(color: Colors.black, fontWeight: FontWeight.bold)),
              ),
            ],
          ),
          const SizedBox(height: 12),
          if (_carteira.isEmpty)
            const Padding(
              padding: EdgeInsets.symmetric(vertical: 30),
              child: Center(child: Text('Nenhum ativo cadastrado. Clique em Adicionar.', style: TextStyle(color: Colors.white54))),
            )
          else
            ...List.generate(_carteira.length, (i) => _buildCardAtivo(_carteira[i], i)),
        ],
      ),
    );
  }

  Widget _buildResumoPatrimonio() {
    return Container(
      padding: const EdgeInsets.all(20),
      decoration: BoxDecoration(
        color: const Color(0xFF161B22),
        borderRadius: BorderRadius.circular(16),
        border: Border.all(color: Colors.white10),
      ),
      child: Column(
        crossAxisAlignment: CrossAxisAlignment.start,
        children: [
          const Text('Patrimônio Total', style: TextStyle(color: Colors.white60, fontSize: 14)),
          const SizedBox(height: 8),
          Text(
            _ocultarValores ? 'R\$ ••••••••' : 'R\$ ${patrimonioTotal.toStringAsFixed(2)}',
            style: const TextStyle(fontSize: 30, fontWeight: FontWeight.bold, color: Colors.white),
          ),
          const SizedBox(height: 16),
          Row(
            mainAxisAlignment: MainAxisAlignment.spaceBetween,
            children: [
              Column(
                crossAxisAlignment: CrossAxisAlignment.start,
                children: [
                  const Text('Rentabilidade', style: TextStyle(color: Colors.white54, fontSize: 12)),
                  Text(
                    '${rentabilidadePct >= 0 ? "+" : ""}${rentabilidadePct.toStringAsFixed(2)}%',
                    style: TextStyle(
                      color: rentabilidadePct >= 0 ? const Color(0xFF00E676) : Colors.redAccent,
                      fontWeight: FontWeight.bold,
                      fontSize: 16,
                    ),
                  ),
                ],
              ),
              Column(
                crossAxisAlignment: CrossAxisAlignment.end,
                children: [
                  const Text('Lucro Total', style: TextStyle(color: Colors.white54, fontSize: 12)),
                  Text(
                    _ocultarValores ? 'R\$ •••' : 'R\$ ${lucroTotal.toStringAsFixed(2)}',
                    style: TextStyle(
                      color: lucroTotal >= 0 ? const Color(0xFF00E676) : Colors.redAccent,
                      fontWeight: FontWeight.bold,
                      fontSize: 16,
                    ),
                  ),
                ],
              ),
            ],
          ),
        ],
      ),
    );
  }

  Widget _buildGraficoAlocacao() {
    if (patrimonioTotal == 0) return const SizedBox.shrink();

    return Container(
      height: 200,
      padding: const EdgeInsets.all(16),
      decoration: BoxDecoration(
        color: const Color(0xFF161B22),
        borderRadius: BorderRadius.circular(16),
        border: Border.all(color: Colors.white10),
      ),
      child: Row(
        children: [
          Expanded(
            child: PieChart(
              PieChartData(
                sectionsSpace: 4,
                centerSpaceRadius: 40,
                sections: _carteira.map((ativo) {
                  final pct = (ativo.valorTotal / patrimonioTotal) * 100;
                  return PieChartSectionData(
                    value: ativo.valorTotal,
                    title: '${pct.toStringAsFixed(0)}%',
                    radius: 35,
                    titleStyle: const TextStyle(fontSize: 12, fontWeight: FontWeight.bold, color: Colors.white),
                  );
                }).toList(),
              ),
            ),
          ),
          Column(
            mainAxisAlignment: MainAxisAlignment.center,
            crossAxisAlignment: CrossAxisAlignment.start,
            children: _carteira.map((a) => Padding(
              padding: const EdgeInsets.symmetric(vertical: 4),
              child: Row(
                children: [
                  Text('${a.ticker}: ', style: const TextStyle(fontWeight: FontWeight.bold)),
                  Text('${((a.valorTotal / patrimonioTotal) * 100).toStringAsFixed(1)}%', style: const TextStyle(color: Colors.white70)),
                ],
              ),
            )).toList(),
          )
        ],
      ),
    );
  }

  Widget _buildCardAtivo(AtivoModel ativo, int index) {
    final pesoPct = patrimonioTotal > 0 ? (ativo.valorTotal / patrimonioTotal) * 100 : 0.0;
    final variacaoPositiva = ativo.variacaoDia >= 0;

    return Dismissible(
      key: Key(ativo.ticker + index.toString()),
      direction: DismissDirection.endToStart,
      background: Container(
        alignment: Alignment.centerRight,
        padding: const EdgeInsets.only(right: 20),
        decoration: BoxDecoration(color: Colors.redAccent, borderRadius: BorderRadius.circular(12)),
        child: const Icon(Icons.delete, color: Colors.white),
      ),
      onDismissed: (_) => _removerAtivo(index),
      child: Card(
        margin: const EdgeInsets.only(bottom: 12),
        shape: RoundedRectangleBorder(borderRadius: BorderRadius.circular(12)),
        child: ListTile(
          contentPadding: const EdgeInsets.symmetric(horizontal: 16, vertical: 8),
          leading: CircleAvatar(
            backgroundColor: const Color(0xFF00E676).withOpacity(0.1),
            child: Text(ativo.ticker.substring(0, 2), style: const TextStyle(color: Color(0xFF00E676), fontWeight: FontWeight.bold)),
          ),
          title: Row(
            children: [
              Text(ativo.ticker, style: const TextStyle(fontWeight: FontWeight.bold)),
              const SizedBox(width: 8),
              Container(
                padding: const EdgeInsets.symmetric(horizontal: 6, vertical: 2),
                decoration: BoxDecoration(color: Colors.white10, borderRadius: BorderRadius.circular(4)),
                child: Text(ativo.classe, style: const TextStyle(fontSize: 10, color: Colors.white70)),
              )
            ],
          ),
          subtitle: Text('Peso: ${pesoPct.toStringAsFixed(1)}% | Qtd: ${ativo.quantidade.toInt()}', style: const TextStyle(color: Colors.white54, fontSize: 12)),
          trailing: Column(
            mainAxisAlignment: MainAxisAlignment.center,
            crossAxisAlignment: CrossAxisAlignment.end,
            children: [
              Text(
                _ocultarValores ? 'R\$ •••' : 'R\$ ${ativo.precoAtual.toStringAsFixed(2)}',
                style: const TextStyle(fontWeight: FontWeight.bold, fontSize: 15),
              ),
              Text(
                '${variacaoPositiva ? "+" : ""}${ativo.variacaoDia.toStringAsFixed(2)}%',
                style: TextStyle(
                  color: variacaoPositiva ? const Color(0xFF00E676) : Colors.redAccent,
                  fontWeight: FontWeight.bold,
                  fontSize: 12,
                ),
              ),
            ],
          ),
        ),
      ),
    );
  }
}
```

```python
%%writefile /content/investidor_app/lib/main.dart
import 'package:flutter/material.dart';
import 'package:fl_chart/fl_chart.dart';
import 'dart:convert';
import 'dart:math' as math;
import 'package:http/http.dart' as http;
import 'package:shared_preferences/shared_preferences.dart';

void main() {
  runApp(const InvestidorApp());
}

class AtivoModel {
  final String ticker;
  final String nome;
  final String classe;
  final String setor;
  double quantidade;
  double precoMedio;
  double precoAtual;
  double variacaoDia;
  double pl;
  double dy;
  double pvp;
  double roe;
  double lpa;
  double vpa;

  AtivoModel({
    required this.ticker,
    required this.nome,
    required this.classe,
    required this.setor,
    required this.quantidade,
    required this.precoMedio,
    this.precoAtual = 0.0,
    this.variacaoDia = 0.0,
    this.pl = 5.4,
    this.dy = 9.8,
    this.pvp = 1.12,
    this.roe = 22.5,
    this.lpa = 6.20,
    this.vpa = 32.10,
  });

  double get valorTotal => quantidade * (precoAtual > 0 ? precoAtual : precoMedio);
  double get totalInvestido => quantidade * precoMedio;
  double get lucroPrejuizo => valorTotal - totalInvestido;
  double get rentabilidadePct => totalInvestido > 0 ? (lucroPrejuizo / totalInvestido) * 100 : 0.0;

  Map<String, dynamic> toMap() {
    return {
      'ticker': ticker,
      'nome': nome,
      'classe': classe,
      'setor': setor,
      'quantidade': quantidade,
      'precoMedio': precoMedio,
      'pl': pl,
      'dy': dy,
      'pvp': pvp,
      'roe': roe,
      'lpa': lpa,
      'vpa': vpa,
    };
  }

  factory AtivoModel.fromMap(Map<String, dynamic> map) {
    return AtivoModel(
      ticker: map['ticker'] ?? '',
      nome: map['nome'] ?? map['ticker'] ?? '',
      classe: map['classe'] ?? 'Ações',
      setor: map['setor'] ?? 'Geral',
      quantidade: (map['quantidade'] as num).toDouble(),
      precoMedio: (map['precoMedio'] as num).toDouble(),
      precoAtual: (map['precoMedio'] as num).toDouble(),
      pl: (map['pl'] as num?)?.toDouble() ?? 5.4,
      dy: (map['dy'] as num?)?.toDouble() ?? 9.8,
      pvp: (map['pvp'] as num?)?.toDouble() ?? 1.12,
      roe: (map['roe'] as num?)?.toDouble() ?? 22.5,
      lpa: (map['lpa'] as num?)?.toDouble() ?? 6.20,
      vpa: (map['vpa'] as num?)?.toDouble() ?? 32.10,
    );
  }
}

class InvestidorApp extends StatelessWidget {
  const InvestidorApp({super.key});

  @override
  Widget build(BuildContext context) {
    return MaterialApp(
      debugShowCheckedModeBanner: false,
      title: 'Investidor IA',
      theme: ThemeData.dark().copyWith(
        scaffoldBackgroundColor: const Color(0xFF0D1117),
        cardColor: const Color(0xFF161B22),
        primaryColor: const Color(0xFF00E676),
      ),
      home: const DashboardScreen(),
    );
  }
}

class DashboardScreen extends StatefulWidget {
  const DashboardScreen({super.key});

  @override
  State<DashboardScreen> createState() => _DashboardScreenState();
}

class _DashboardScreenState extends State<DashboardScreen> {
  bool _ocultarValores = false;
  bool _carregandoCotacoes = false;
  List<AtivoModel> _carteira = [];

  @override
  void initState() {
    super.initState();
    _carregarCarteiraLocal();
  }

  Future<void> _salvarCarteiraLocal() async {
    final prefs = await SharedPreferences.getInstance();
    final List<String> carteiraJson = _carteira.map((item) => json.encode(item.toMap())).toList();
    await prefs.setStringList('minha_carteira', carteiraJson);
  }

  Future<void> _carregarCarteiraLocal() async {
    final prefs = await SharedPreferences.getInstance();
    final List<String>? carteiraJson = prefs.getStringList('minha_carteira');

    if (carteiraJson != null && carteiraJson.isNotEmpty) {
      setState(() {
        _carteira = carteiraJson.map((str) => AtivoModel.fromMap(json.decode(str))).toList();
      });
    } else {
      setState(() {
        _carteira = [
          AtivoModel(ticker: 'PETR4', nome: 'Petrobras PN', classe: 'Ações', setor: 'Energia', quantidade: 150, precoMedio: 32.50, precoAtual: 38.50, variacaoDia: 1.45, pl: 4.2, dy: 14.8, pvp: 1.15, roe: 28.5, lpa: 9.15, vpa: 33.40),
          AtivoModel(ticker: 'VALE3', nome: 'Vale ON', classe: 'Ações', setor: 'Mineração', quantidade: 80, precoMedio: 68.00, precoAtual: 62.10, variacaoDia: -0.85, pl: 5.8, dy: 8.2, pvp: 1.32, roe: 21.0, lpa: 10.70, vpa: 47.00),
          AtivoModel(ticker: 'MXRF11', nome: 'Maxi Renda FII', classe: 'FIIs', setor: 'Papel', quantidade: 500, precoMedio: 10.10, precoAtual: 10.45, variacaoDia: 0.10, pl: 0.0, dy: 12.4, pvp: 1.02, roe: 0.0, lpa: 1.10, vpa: 10.25),
        ];
      });
      _salvarCarteiraLocal();
    }
    _atualizarCotacoes();
  }

  double get patrimonioTotal => _carteira.fold(0, (sum, item) => sum + item.valorTotal);
  double get totalInvestido => _carteira.fold(0, (sum, item) => sum + item.totalInvestido);
  double get lucroTotal => patrimonioTotal - totalInvestido;
  double get rentabilidadePct => totalInvestido > 0 ? (lucroTotal / totalInvestido) * 100 : 0;

  Future<void> _atualizarCotacoes() async {
    if (_carteira.isEmpty) return;
    setState(() => _carregandoCotacoes = true);

    for (var ativo in _carteira) {
      try {
        final url = Uri.parse('https://brapi.dev/api/quote/${ativo.ticker}');
        final response = await http.get(url);
        if (response.statusCode == 200) {
          final data = json.decode(response.body);
          if (data['results'] != null && (data['results'] as List).isNotEmpty) {
            final result = data['results'][0];
            setState(() {
              ativo.precoAtual = (result['regularMarketPrice'] as num).toDouble();
              ativo.variacaoDia = (result['regularMarketChangePercent'] as num).toDouble();
            });
          }
        }
      } catch (e) {
        print('Erro no ticker ${ativo.ticker}: $e');
      }
    }
    setState(() => _carregandoCotacoes = false);
  }

  void _removerAtivo(int index) {
    setState(() {
      _carteira.removeAt(index);
    });
    _salvarCarteiraLocal();
  }

  void _exibirModalAdicionarAtivo() {
    final tickerCtrl = TextEditingController();
    final qtdCtrl = TextEditingController();
    final precoCtrl = TextEditingController();
    String classeSel = 'Ações';

    showModalBottomSheet(
      context: context,
      isScrollControlled: true,
      backgroundColor: const Color(0xFF161B22),
      shape: const RoundedRectangleBorder(
        borderRadius: BorderRadius.vertical(top: Radius.circular(20)),
      ),
      builder: (ctx) => StatefulBuilder(
        builder: (ctxState, setModalState) => Padding(
          padding: EdgeInsets.only(
            top: 20, left: 20, right: 20,
            bottom: MediaQuery.of(ctx).viewInsets.bottom + 20,
          ),
          child: Column(
            mainAxisSize: MainAxisSize.min,
            crossAxisAlignment: CrossAxisAlignment.start,
            children: [
              const Text('Adicionar Novo Ativo', style: TextStyle(fontSize: 18, fontWeight: FontWeight.bold)),
              const SizedBox(height: 16),
              TextField(
                controller: tickerCtrl,
                textCapitalization: TextCapitalization.characters,
                decoration: const InputDecoration(
                  labelText: 'Ticker (ex: WEGE3, HGLG11)',
                  border: OutlineInputBorder(),
                ),
              ),
              const SizedBox(height: 12),
              DropdownButtonFormField<String>(
                value: classeSel,
                decoration: const InputDecoration(labelText: 'Classe', border: OutlineInputBorder()),
                items: ['Ações', 'FIIs', 'Renda Fixa', 'BDRs'].map((c) => DropdownMenuItem(value: c, child: Text(c))).toList(),
                onChanged: (val) => setModalState(() => classeSel = val!),
              ),
              const SizedBox(height: 12),
              Row(
                children: [
                  Expanded(
                    child: TextField(
                      controller: qtdCtrl,
                      keyboardType: TextInputType.number,
                      decoration: const InputDecoration(labelText: 'Quantidade', border: OutlineInputBorder()),
                    ),
                  ),
                  const SizedBox(width: 12),
                  Expanded(
                    child: TextField(
                      controller: precoCtrl,
                      keyboardType: const TextInputType.numberWithOptions(decimal: true),
                      decoration: const InputDecoration(labelText: 'Preço Médio (R\$)', border: OutlineInputBorder()),
                    ),
                  ),
                ],
              ),
              const SizedBox(height: 20),
              ElevatedButton(
                style: ElevatedButton.styleFrom(
                  backgroundColor: const Color(0xFF00E676),
                  minimumSize: const Size.fromHeight(50),
                ),
                onPressed: () {
                  final ticker = tickerCtrl.text.trim().toUpperCase();
                  final qtd = double.tryParse(qtdCtrl.text) ?? 0;
                  final preco = double.tryParse(precoCtrl.text.replaceAll(',', '.')) ?? 0;

                  if (ticker.isNotEmpty && qtd > 0 && preco > 0) {
                    final novoAtivo = AtivoModel(
                      ticker: ticker,
                      nome: ticker,
                      classe: classeSel,
                      setor: 'Geral',
                      quantidade: qtd,
                      precoMedio: preco,
                      precoAtual: preco,
                    );
                    setState(() {
                      _carteira.add(novoAtivo);
                    });
                    _salvarCarteiraLocal();
                    Navigator.pop(ctx);
                    _atualizarCotacoes();
                  }
                },
                child: const Text('Salvar Ativo', style: TextStyle(color: Colors.black, fontWeight: FontWeight.bold)),
              )
            ],
          ),
        ),
      ),
    );
  }

  @override
  Widget build(BuildContext context) {
    return Scaffold(
      appBar: AppBar(
        backgroundColor: const Color(0xFF161B22),
        elevation: 0,
        title: const Text('Investidor IA', style: TextStyle(fontWeight: FontWeight.bold)),
        actions: [
          IconButton(
            icon: Icon(_ocultarValores ? Icons.visibility_off : Icons.visibility),
            onPressed: () => setState(() => _ocultarValores = !_ocultarValores),
          ),
          IconButton(
            icon: _carregandoCotacoes
                ? const SizedBox(width: 20, height: 20, child: CircularProgressIndicator(strokeWidth: 2))
                : const Icon(Icons.refresh),
            onPressed: _atualizarCotacoes,
          ),
        ],
      ),
      body: ListView(
        padding: const EdgeInsets.all(16),
        children: [
          _buildResumoPatrimonio(),
          const SizedBox(height: 20),
          _buildGraficoAlocacao(),
          const SizedBox(height: 24),
          Row(
            mainAxisAlignment: MainAxisAlignment.spaceBetween,
            children: [
              const Text('Meus Ativos', style: TextStyle(fontSize: 18, fontWeight: FontWeight.bold)),
              ElevatedButton.icon(
                style: ElevatedButton.styleFrom(backgroundColor: const Color(0xFF00E676)),
                onPressed: _exibirModalAdicionarAtivo,
                icon: const Icon(Icons.add, color: Colors.black),
                label: const Text('Adicionar', style: TextStyle(color: Colors.black, fontWeight: FontWeight.bold)),
              ),
            ],
          ),
          const SizedBox(height: 12),
          if (_carteira.isEmpty)
            const Padding(
              padding: EdgeInsets.symmetric(vertical: 30),
              child: Center(child: Text('Nenhum ativo cadastrado. Clique em Adicionar.', style: TextStyle(color: Colors.white54))),
            )
          else
            ...List.generate(_carteira.length, (i) => _buildCardAtivo(_carteira[i], i)),
        ],
      ),
    );
  }

  Widget _buildResumoPatrimonio() {
    return Container(
      padding: const EdgeInsets.all(20),
      decoration: BoxDecoration(
        color: const Color(0xFF161B22),
        borderRadius: BorderRadius.circular(16),
        border: Border.all(color: Colors.white10),
      ),
      child: Column(
        crossAxisAlignment: CrossAxisAlignment.start,
        children: [
          const Text('Patrimônio Total', style: TextStyle(color: Colors.white60, fontSize: 14)),
          const SizedBox(height: 8),
          Text(
            _ocultarValores ? 'R\$ ••••••••' : 'R\$ ${patrimonioTotal.toStringAsFixed(2)}',
            style: const TextStyle(fontSize: 30, fontWeight: FontWeight.bold, color: Colors.white),
          ),
          const SizedBox(height: 16),
          Row(
            mainAxisAlignment: MainAxisAlignment.spaceBetween,
            children: [
              Column(
                crossAxisAlignment: CrossAxisAlignment.start,
                children: [
                  const Text('Rentabilidade', style: TextStyle(color: Colors.white54, fontSize: 12)),
                  Text(
                    '${rentabilidadePct >= 0 ? "+" : ""}${rentabilidadePct.toStringAsFixed(2)}%',
                    style: TextStyle(
                      color: rentabilidadePct >= 0 ? const Color(0xFF00E676) : Colors.redAccent,
                      fontWeight: FontWeight.bold,
                      fontSize: 16,
                    ),
                  ),
                ],
              ),
              Column(
                crossAxisAlignment: CrossAxisAlignment.end,
                children: [
                  const Text('Lucro Total', style: TextStyle(color: Colors.white54, fontSize: 12)),
                  Text(
                    _ocultarValores ? 'R\$ •••' : 'R\$ ${lucroTotal.toStringAsFixed(2)}',
                    style: TextStyle(
                      color: lucroTotal >= 0 ? const Color(0xFF00E676) : Colors.redAccent,
                      fontWeight: FontWeight.bold,
                      fontSize: 16,
                    ),
                  ),
                ],
              ),
            ],
          ),
        ],
      ),
    );
  }

  Widget _buildGraficoAlocacao() {
    if (patrimonioTotal == 0) return const SizedBox.shrink();

    return Container(
      height: 200,
      padding: const EdgeInsets.all(16),
      decoration: BoxDecoration(
        color: const Color(0xFF161B22),
        borderRadius: BorderRadius.circular(16),
        border: Border.all(color: Colors.white10),
      ),
      child: Row(
        children: [
          Expanded(
            child: PieChart(
              PieChartData(
                sectionsSpace: 4,
                centerSpaceRadius: 40,
                sections: _carteira.map((ativo) {
                  final pct = (ativo.valorTotal / patrimonioTotal) * 100;
                  return PieChartSectionData(
                    value: ativo.valorTotal,
                    title: '${pct.toStringAsFixed(0)}%',
                    radius: 35,
                    titleStyle: const TextStyle(fontSize: 12, fontWeight: FontWeight.bold, color: Colors.white),
                  );
                }).toList(),
              ),
            ),
          ),
          Column(
            mainAxisAlignment: MainAxisAlignment.center,
            crossAxisAlignment: CrossAxisAlignment.start,
            children: _carteira.map((a) => Padding(
              padding: const EdgeInsets.symmetric(vertical: 4),
              child: Row(
                children: [
                  Text('${a.ticker}: ', style: const TextStyle(fontWeight: FontWeight.bold)),
                  Text('${((a.valorTotal / patrimonioTotal) * 100).toStringAsFixed(1)}%', style: const TextStyle(color: Colors.white70)),
                ],
              ),
            )).toList(),
          )
        ],
      ),
    );
  }

  Widget _buildCardAtivo(AtivoModel ativo, int index) {
    final pesoPct = patrimonioTotal > 0 ? (ativo.valorTotal / patrimonioTotal) * 100 : 0.0;
    final variacaoPositiva = ativo.variacaoDia >= 0;

    return Dismissible(
      key: Key(ativo.ticker + index.toString()),
      direction: DismissDirection.endToStart,
      background: Container(
        alignment: Alignment.centerRight,
        padding: const EdgeInsets.only(right: 20),
        decoration: BoxDecoration(color: Colors.redAccent, borderRadius: BorderRadius.circular(12)),
        child: const Icon(Icons.delete, color: Colors.white),
      ),
      onDismissed: (_) => _removerAtivo(index),
      child: Card(
        margin: const EdgeInsets.only(bottom: 12),
        shape: RoundedRectangleBorder(borderRadius: BorderRadius.circular(12)),
        child: ListTile(
          contentPadding: const EdgeInsets.symmetric(horizontal: 16, vertical: 8),
          onTap: () {
            Navigator.push(
              context,
              MaterialPageRoute(
                builder: (context) => DetalhesAtivoScreen(ativo: ativo),
              ),
            );
          },
          leading: CircleAvatar(
            backgroundColor: const Color(0xFF00E676).withOpacity(0.1),
            child: Text(ativo.ticker.substring(0, 2), style: const TextStyle(color: Color(0xFF00E676), fontWeight: FontWeight.bold)),
          ),
          title: Row(
            children: [
              Text(ativo.ticker, style: const TextStyle(fontWeight: FontWeight.bold)),
              const SizedBox(width: 8),
              Container(
                padding: const EdgeInsets.symmetric(horizontal: 6, vertical: 2),
                decoration: BoxDecoration(color: Colors.white10, borderRadius: BorderRadius.circular(4)),
                child: Text(ativo.classe, style: const TextStyle(fontSize: 10, color: Colors.white70)),
              )
            ],
          ),
          subtitle: Text('Peso: ${pesoPct.toStringAsFixed(1)}% | Qtd: ${ativo.quantidade.toInt()}', style: const TextStyle(color: Colors.white54, fontSize: 12)),
          trailing: Column(
            mainAxisAlignment: MainAxisAlignment.center,
            crossAxisAlignment: CrossAxisAlignment.end,
            children: [
              Text(
                _ocultarValores ? 'R\$ •••' : 'R\$ ${ativo.precoAtual.toStringAsFixed(2)}',
                style: const TextStyle(fontWeight: FontWeight.bold, fontSize: 15),
              ),
              Text(
                '${variacaoPositiva ? "+" : ""}${ativo.variacaoDia.toStringAsFixed(2)}%',
                style: TextStyle(
                  color: variacaoPositiva ? const Color(0xFF00E676) : Colors.redAccent,
                  fontWeight: FontWeight.bold,
                  fontSize: 12,
                ),
              ),
            ],
          ),
        ),
      ),
    );
  }
}

class DetalhesAtivoScreen extends StatelessWidget {
  final AtivoModel ativo;

  const DetalhesAtivoScreen({super.key, required this.ativo});

  double get precoJustoGraham {
    if (ativo.lpa > 0 && ativo.vpa > 0) {
      return math.sqrt(22.5 * ativo.lpa * ativo.vpa);
    }
    return 0.0;
  }

  double get precoTetoBarsi {
    if (ativo.dy > 0 && ativo.precoAtual > 0) {
      final proventoEstimado12M = (ativo.dy / 100) * ativo.precoAtual;
      return proventoEstimado12M / 0.06;
    }
    return 0.0;
  }

  @override
  Widget build(BuildContext context) {
    final variacaoPositiva = ativo.variacaoDia >= 0;
    final grahamValido = precoJustoGraham > 0;
    final descontoGraham = grahamValido ? ((precoJustoGraham - ativo.precoAtual) / precoJustoGraham) * 100 : 0.0;

    return Scaffold(
      appBar: AppBar(
        backgroundColor: const Color(0xFF161B22),
        elevation: 0,
        title: Text('${ativo.ticker} - Ficha Técnica', style: const TextStyle(fontWeight: FontWeight.bold)),
      ),
      body: ListView(
        padding: const EdgeInsets.all(16),
        children: [
          // Header principal do ativo
          Container(
            padding: const EdgeInsets.all(20),
            decoration: BoxDecoration(
              color: const Color(0xFF161B22),
              borderRadius: BorderRadius.circular(16),
              border: Border.all(color: Colors.white10),
            ),
            child: Row(
              mainAxisAlignment: MainAxisAlignment.spaceBetween,
              children: [
                Column(
                  crossAxisAlignment: CrossAxisAlignment.start,
                  children: [
                    Text(ativo.nome, style: const TextStyle(color: Colors.white54, fontSize: 14)),
                    const SizedBox(height: 4),
                    Text(
                      'R\$ ${ativo.precoAtual.toStringAsFixed(2)}',
                      style: const TextStyle(fontSize: 28, fontWeight: FontWeight.bold, color: Colors.white),
                    ),
                  ],
                ),
                Container(
                  padding: const EdgeInsets.symmetric(horizontal: 12, vertical: 6),
                  decoration: BoxDecoration(
                    color: variacaoPositiva ? Colors.green.withOpacity(0.2) : Colors.red.withOpacity(0.2),
                    borderRadius: BorderRadius.circular(8),
                  ),
                  child: Text(
                    '${variacaoPositiva ? "+" : ""}${ativo.variacaoDia.toStringAsFixed(2)}%',
                    style: TextStyle(
                      color: variacaoPositiva ? const Color(0xFF00E676) : Colors.redAccent,
                      fontWeight: FontWeight.bold,
                      fontSize: 14,
                    ),
                  ),
                )
              ],
            ),
          ),
          const SizedBox(height: 20),

          // Card de Valuation (Graham / Barsi)
          Container(
            padding: const EdgeInsets.all(16),
            decoration: BoxDecoration(
              color: const Color(0xFF161B22),
              borderRadius: BorderRadius.circular(16),
              border: Border.all(color: Colors.white10),
            ),
            child: Column(
              crossAxisAlignment: CrossAxisAlignment.start,
              children: [
                const Text('Valuation & Margem de Segurança', style: TextStyle(fontSize: 16, fontWeight: FontWeight.bold)),
                const SizedBox(height: 12),
                Row(
                  mainAxisAlignment: MainAxisAlignment.spaceBetween,
                  children: [
                    _buildValuationMetric('Preço Justo (Graham)', grahamValido ? 'R\$ ${precoJustoGraham.toStringAsFixed(2)}' : 'N/A', descontoGraham >= 0 ? '+${descontoGraham.toStringAsFixed(1)}% Desconto' : '${descontoGraham.toStringAsFixed(1)}% Sobrepreço'),
                    _buildValuationMetric('Preço Teto 6% (Barsi)', precoTetoBarsi > 0 ? 'R\$ ${precoTetoBarsi.toStringAsFixed(2)}' : 'N/A', 'Margem 6% DY'),
                  ],
                )
              ],
            ),
          ),
          const SizedBox(height: 20),

          // Gráfico de Cotação Histórica
          const Text('Cotação Histórica (12 Meses)', style: TextStyle(fontSize: 16, fontWeight: FontWeight.bold)),
          const SizedBox(height: 12),
          Container(
            height: 180,
            padding: const EdgeInsets.only(right: 16, top: 16, bottom: 8),
            decoration: BoxDecoration(
              color: const Color(0xFF161B22),
              borderRadius: BorderRadius.circular(16),
              border: Border.all(color: Colors.white10),
            ),
            child: LineChart(
              LineChartData(
                gridData: FlGridData(show: false),
                titlesData: FlTitlesData(show: false),
                borderData: FlBorderData(show: false),
                lineBarsData: [
                  LineChartBarData(
                    spots: const [
                      FlSpot(0, 30),
                      FlSpot(2, 32),
                      FlSpot(4, 28),
                      FlSpot(6, 35),
                      FlSpot(8, 34),
                      FlSpot(10, 38.5),
                    ],
                    isCurved: true,
                    color: const Color(0xFF00E676),
                    barWidth: 3,
                    dotData: FlDotData(show: false),
                    belowBarData: BarAreaData(
                      show: true,
                      color: const Color(0xFF00E676).withOpacity(0.15),
                    ),
                  ),
                ],
              ),
            ),
          ),
          const SizedBox(height: 24),

          // Grid de Indicadores Fundamentalistas
          const Text('Indicadores Fundamentalistas', style: TextStyle(fontSize: 16, fontWeight: FontWeight.bold)),
          const SizedBox(height: 12),
          GridView.count(
            crossAxisCount: 2,
            shrinkWrap: true,
            physics: const NeverScrollableScrollPhysics(),
            childAspectRatio: 2.1,
            crossAxisSpacing: 12,
            mainAxisSpacing: 12,
            children: [
              _buildCardIndicador('P/L (Preço / Lucro)', ativo.pl > 0 ? ativo.pl.toStringAsFixed(2) : '-'),
              _buildCardIndicador('Dividend Yield (DY 12M)', '${ativo.dy.toStringAsFixed(2)}%'),
              _buildCardIndicador('P/VP (Preço / V. Patr.)', ativo.pvp.toStringAsFixed(2)),
              _buildCardIndicador('ROE (Retorno s/ Patr.)', '${ativo.roe.toStringAsFixed(2)}%'),
              _buildCardIndicador('LPA (Lucro por Ação)', 'R\$ ${ativo.lpa.toStringAsFixed(2)}'),
              _buildCardIndicador('VPA (Val. Patr. por Ação)', 'R\$ ${ativo.vpa.toStringAsFixed(2)}'),
            ],
          ),
          const SizedBox(height: 24),

          // Extrato de Proventos Recentes
          const Text('Histórico de Proventos', style: TextStyle(fontSize: 16, fontWeight: FontWeight.bold)),
          const SizedBox(height: 12),
          _buildItemProvento('Dividendo', '15/10/2024', 'R\$ 0,65 por ação'),
          _buildItemProvento('JCP', '12/07/2024', 'R\$ 0,42 por ação'),
          _buildItemProvento('Dividendo', '20/04/2024', 'R\$ 0,88 por ação'),
        ],
      ),
    );
  }

  Widget _buildValuationMetric(String titulo, String valor, String status) {
    return Column(
      crossAxisAlignment: CrossAxisAlignment.start,
      children: [
        Text(titulo, style: const TextStyle(color: Colors.white54, fontSize: 12)),
        const SizedBox(height: 4),
        Text(valor, style: const TextStyle(fontSize: 18, fontWeight: FontWeight.bold, color: Color(0xFF00E676))),
        const SizedBox(height: 2),
        Text(status, style: const TextStyle(color: Colors.white70, fontSize: 11)),
      ],
    );
  }

  Widget _buildCardIndicador(String titulo, String valor) {
    return Container(
      padding: const EdgeInsets.all(12),
      decoration: BoxDecoration(
        color: const Color(0xFF161B22),
        borderRadius: BorderRadius.circular(12),
        border: Border.all(color: Colors.white10),
      ),
      child: Column(
        crossAxisAlignment: CrossAxisAlignment.start,
        mainAxisAlignment: MainAxisAlignment.center,
        children: [
          Text(titulo, style: const TextStyle(color: Colors.white54, fontSize: 11)),
          const SizedBox(height: 6),
          Text(valor, style: const TextStyle(fontSize: 16, fontWeight: FontWeight.bold, color: Colors.white)),
        ],
      ),
    );
  }

  Widget _buildItemProvento(String tipo, String data, String valorText) {
    return Container(
      margin: const EdgeInsets.only(bottom: 8),
      padding: const EdgeInsets.all(12),
      decoration: BoxDecoration(
        color: const Color(0xFF161B22),
        borderRadius: BorderRadius.circular(10),
        border: Border.all(color: Colors.white10),
      ),
      child: Row(
        mainAxisAlignment: MainAxisAlignment.spaceBetween,
        children: [
          Row(
            children: [
              Icon(tipo == 'Dividendo' ? Icons.attach_money : Icons.account_balance_wallet, color: const Color(0xFF00E676), size: 20),
              const SizedBox(width: 10),
              Column(
                crossAxisAlignment: CrossAxisAlignment.start,
                children: [
                  Text(tipo, style: const TextStyle(fontWeight: FontWeight.bold)),
                  Text('Data com: $data', style: const TextStyle(color: Colors.white54, fontSize: 11)),
                ],
              )
            ],
          ),
          Text(valorText, style: const TextStyle(fontWeight: FontWeight.bold, color: Colors.white70, fontSize: 13)),
        ],
      ),
    );
  }
}
```

```python
%%writefile /content/investidor_app/lib/main.dart
import 'package:flutter/material.dart';
import 'package:fl_chart/fl_chart.dart';
import 'dart:convert';
import 'dart:math' as math;
import 'package:http/http.dart' as http;
import 'package:firebase_core/firebase_core.dart';
import 'package:firebase_auth/firebase_auth.dart';
import 'package:cloud_firestore/cloud_firestore.dart';

void main() async {
  WidgetsFlutterBinding.ensureInitialized();
  try {
    await Firebase.initializeApp();
  } catch (e) {
    print('Erro ao inicializar Firebase: $e');
  }
  runApp(const InvestidorApp());
}

class AuthService {
  final FirebaseAuth _auth = FirebaseAuth.instance;

  Stream<User?> get authStateChanges => _auth.authStateChanges();
  String? get usuarioId => _auth.currentUser?.uid;

  Future<UserCredential?> cadastrar(String email, String senha) async {
    return await _auth.createUserWithEmailAndPassword(email: email, password: senha);
  }

  Future<UserCredential?> entrar(String email, String senha) async {
    return await _auth.signInWithEmailAndPassword(email: email, password: senha);
  }

  Future<void> sair() async {
    await _auth.signOut();
  }
}

class FirestoreService {
  final FirebaseFirestore _db = FirebaseFirestore.instance;

  Future<void> salvarCarteiraNuvem(String uid, List<AtivoModel> carteira) async {
    final batch = _db.batch();
    final colecaoRef = _db.collection('usuarios').doc(uid).collection('carteira');

    final snapshots = await colecaoRef.get();
    for (var doc in snapshots.docs) {
      batch.delete(doc.reference);
    }

    for (var ativo in carteira) {
      final docRef = colecaoRef.doc(ativo.ticker);
      batch.set(docRef, ativo.toMap());
    }

    await batch.commit();
  }

  Future<List<AtivoModel>> carregarCarteiraNuvem(String uid) async {
    final snapshot = await _db.collection('usuarios').doc(uid).collection('carteira').get();
    return snapshot.docs.map((doc) => AtivoModel.fromMap(doc.data())).toList();
  }
}

class AtivoModel {
  final String ticker;
  final String nome;
  final String classe;
  final String setor;
  double quantidade;
  double precoMedio;
  double precoAtual;
  double variacaoDia;
  double pl;
  double dy;
  double pvp;
  double roe;
  double lpa;
  double vpa;

  AtivoModel({
    required this.ticker,
    required this.nome,
    required this.classe,
    required this.setor,
    required this.quantidade,
    required this.precoMedio,
    this.precoAtual = 0.0,
    this.variacaoDia = 0.0,
    this.pl = 5.4,
    this.dy = 9.8,
    this.pvp = 1.12,
    this.roe = 22.5,
    this.lpa = 6.20,
    this.vpa = 32.10,
  });

  double get valorTotal => quantidade * (precoAtual > 0 ? precoAtual : precoMedio);
  double get totalInvestido => quantidade * precoMedio;
  double get lucroPrejuizo => valorTotal - totalInvestido;
  double get rentabilidadePct => totalInvestido > 0 ? (lucroPrejuizo / totalInvestido) * 100 : 0.0;

  Map<String, dynamic> toMap() {
    return {
      'ticker': ticker,
      'nome': nome,
      'classe': classe,
      'setor': setor,
      'quantidade': quantidade,
      'precoMedio': precoMedio,
      'pl': pl,
      'dy': dy,
      'pvp': pvp,
      'roe': roe,
      'lpa': lpa,
      'vpa': vpa,
    };
  }

  factory AtivoModel.fromMap(Map<String, dynamic> map) {
    return AtivoModel(
      ticker: map['ticker'] ?? '',
      nome: map['nome'] ?? map['ticker'] ?? '',
      classe: map['classe'] ?? 'Ações',
      setor: map['setor'] ?? 'Geral',
      quantidade: (map['quantidade'] as num).toDouble(),
      precoMedio: (map['precoMedio'] as num).toDouble(),
      precoAtual: (map['precoMedio'] as num).toDouble(),
      pl: (map['pl'] as num?)?.toDouble() ?? 5.4,
      dy: (map['dy'] as num?)?.toDouble() ?? 9.8,
      pvp: (map['pvp'] as num?)?.toDouble() ?? 1.12,
      roe: (map['roe'] as num?)?.toDouble() ?? 22.5,
      lpa: (map['lpa'] as num?)?.toDouble() ?? 6.20,
      vpa: (map['vpa'] as num?)?.toDouble() ?? 32.10,
    );
  }
}

class InvestidorApp extends StatelessWidget {
  const InvestidorApp({super.key});

  @override
  Widget build(BuildContext context) {
    return MaterialApp(
      debugShowCheckedModeBanner: false,
      title: 'Investidor IA',
      theme: ThemeData.dark().copyWith(
        scaffoldBackgroundColor: const Color(0xFF0D1117),
        cardColor: const Color(0xFF161B22),
        primaryColor: const Color(0xFF00E676),
      ),
      home: const MainNavigationShell(),
    );
  }
}

class MainNavigationShell extends StatefulWidget {
  const MainNavigationShell({super.key});

  @override
  State<MainNavigationShell> createState() => _MainNavigationShellState();
}

class _MainNavigationShellState extends State<MainNavigationShell> {
  int _abaAtual = 0;
  List<AtivoModel> _carteira = [];
  final FirestoreService _firestoreService = FirestoreService();
  final AuthService _authService = AuthService();

  @override
  void initState() {
    super.initState();
    _carregarCarteira();
  }

  Future<void> _carregarCarteira() async {
    final uid = _authService.usuarioId;
    if (uid != null) {
      final dadosNuvem = await _firestoreService.carregarCarteiraNuvem(uid);
      if (dadosNuvem.isNotEmpty) {
        setState(() => _carteira = dadosNuvem);
        return;
      }
    }
    setState(() {
      _carteira = [
        AtivoModel(ticker: 'PETR4', nome: 'Petrobras PN', classe: 'Ações', setor: 'Energia', quantidade: 150, precoMedio: 32.50, precoAtual: 38.50, variacaoDia: 1.45),
        AtivoModel(ticker: 'VALE3', nome: 'Vale ON', classe: 'Ações', setor: 'Mineração', quantidade: 80, precoMedio: 68.00, precoAtual: 62.10, variacaoDia: -0.85),
        AtivoModel(ticker: 'MXRF11', nome: 'Maxi Renda FII', classe: 'FIIs', setor: 'Papel', quantidade: 500, precoMedio: 10.10, precoAtual: 10.45, variacaoDia: 0.10),
      ];
    });
  }

  Future<void> _salvarCarteira() async {
    final uid = _authService.usuarioId;
    if (uid != null) {
      await _firestoreService.salvarCarteiraNuvem(uid, _carteira);
    }
  }

  @override
  Widget build(BuildContext context) {
    final telas = [
      DashboardScreen(carteira: _carteira, onCarteiraChanged: _salvarCarteira),
      ProventosScreen(carteira: _carteira),
    ];

    return Scaffold(
      body: IndexedStack(
        index: _abaAtual,
        children: telas,
      ),
      bottomNavigationBar: BottomNavigationBar(
        currentIndex: _abaAtual,
        onTap: (index) => setState(() => _abaAtual = index),
        backgroundColor: const Color(0xFF161B22),
        selectedItemColor: const Color(0xFF00E676),
        unselectedItemColor: Colors.white54,
        items: const [
          BottomNavigationBarItem(icon: Icon(Icons.pie_chart), label: 'Carteira'),
          BottomNavigationBarItem(icon: Icon(Icons.attach_money), label: 'Proventos'),
        ],
      ),
    );
  }
}

class DashboardScreen extends StatefulWidget {
  final List<AtivoModel> carteira;
  final VoidCallback onCarteiraChanged;

  const DashboardScreen({super.key, required this.carteira, required this.onCarteiraChanged});

  @override
  State<DashboardScreen> createState() => _DashboardScreenState();
}

class _DashboardScreenState extends State<DashboardScreen> {
  bool _ocultarValores = false;
  bool _carregandoCotacoes = false;

  double get patrimonioTotal => widget.carteira.fold(0, (sum, item) => sum + item.valorTotal);
  double get totalInvestido => widget.carteira.fold(0, (sum, item) => sum + item.totalInvestido);
  double get lucroTotal => patrimonioTotal - totalInvestido;
  double get rentabilidadePct => totalInvestido > 0 ? (lucroTotal / totalInvestido) * 100 : 0;

  Future<void> _atualizarCotacoes() async {
    if (widget.carteira.isEmpty) return;
    setState(() => _carregandoCotacoes = true);

    for (var ativo in widget.carteira) {
      try {
        final url = Uri.parse('https://brapi.dev/api/quote/${ativo.ticker}');
        final response = await http.get(url);
        if (response.statusCode == 200) {
          final data = json.decode(response.body);
          if (data['results'] != null && (data['results'] as List).isNotEmpty) {
            final result = data['results'][0];
            setState(() {
              ativo.precoAtual = (result['regularMarketPrice'] as num).toDouble();
              ativo.variacaoDia = (result['regularMarketChangePercent'] as num).toDouble();
            });
          }
        }
      } catch (e) {
        print('Erro no ticker ${ativo.ticker}: $e');
      }
    }
    setState(() => _carregandoCotacoes = false);
  }

  void _removerAtivo(int index) {
    setState(() {
      widget.carteira.removeAt(index);
    });
    widget.onCarteiraChanged();
  }

  void _exibirModalAdicionarAtivo() {
    final tickerCtrl = TextEditingController();
    final qtdCtrl = TextEditingController();
    final precoCtrl = TextEditingController();
    String classeSel = 'Ações';

    showModalBottomSheet(
      context: context,
      isScrollControlled: true,
      backgroundColor: const Color(0xFF161B22),
      shape: const RoundedRectangleBorder(borderRadius: BorderRadius.vertical(top: Radius.circular(20))),
      builder: (ctx) => StatefulBuilder(
        builder: (ctxState, setModalState) => Padding(
          padding: EdgeInsets.only(top: 20, left: 20, right: 20, bottom: MediaQuery.of(ctx).viewInsets.bottom + 20),
          child: Column(
            mainAxisSize: MainAxisSize.min,
            crossAxisAlignment: CrossAxisAlignment.start,
            children: [
              const Text('Adicionar Novo Ativo', style: TextStyle(fontSize: 18, fontWeight: FontWeight.bold)),
              const SizedBox(height: 16),
              TextField(
                controller: tickerCtrl,
                textCapitalization: TextCapitalization.characters,
                decoration: const InputDecoration(labelText: 'Ticker (ex: WEGE3, HGLG11)', border: OutlineInputBorder()),
              ),
              const SizedBox(height: 12),
              DropdownButtonFormField<String>(
                value: classeSel,
                decoration: const InputDecoration(labelText: 'Classe', border: OutlineInputBorder()),
                items: ['Ações', 'FIIs', 'Renda Fixa', 'BDRs'].map((c) => DropdownMenuItem(value: c, child: Text(c))).toList(),
                onChanged: (val) => setModalState(() => classeSel = val!),
              ),
              const SizedBox(height: 12),
              Row(
                children: [
                  Expanded(
                    child: TextField(
                      controller: qtdCtrl,
                      keyboardType: TextInputType.number,
                      decoration: const InputDecoration(labelText: 'Quantidade', border: OutlineInputBorder()),
                    ),
                  ),
                  const SizedBox(width: 12),
                  Expanded(
                    child: TextField(
                      controller: precoCtrl,
                      keyboardType: const TextInputType.numberWithOptions(decimal: true),
                      decoration: const InputDecoration(labelText: 'Preço Médio (R\$)', border: OutlineInputBorder()),
                    ),
                  ),
                ],
              ),
              const SizedBox(height: 20),
              ElevatedButton(
                style: ElevatedButton.styleFrom(backgroundColor: const Color(0xFF00E676), minimumSize: const Size.fromHeight(50)),
                onPressed: () {
                  final ticker = tickerCtrl.text.trim().toUpperCase();
                  final qtd = double.tryParse(qtdCtrl.text) ?? 0;
                  final preco = double.tryParse(precoCtrl.text.replaceAll(',', '.')) ?? 0;

                  if (ticker.isNotEmpty && qtd > 0 && preco > 0) {
                    setState(() {
                      widget.carteira.add(AtivoModel(
                        ticker: ticker,
                        nome: ticker,
                        classe: classeSel,
                        setor: 'Geral',
                        quantidade: qtd,
                        precoMedio: preco,
                        precoAtual: preco,
                      ));
                    });
                    widget.onCarteiraChanged();
                    Navigator.pop(ctx);
                    _atualizarCotacoes();
                  }
                },
                child: const Text('Salvar Ativo', style: TextStyle(color: Colors.black, fontWeight: FontWeight.bold)),
              )
            ],
          ),
        ),
      ),
    );
  }

  @override
  Widget build(BuildContext context) {
    return Scaffold(
      appBar: AppBar(
        backgroundColor: const Color(0xFF161B22),
        elevation: 0,
        title: const Text('Investidor IA', style: TextStyle(fontWeight: FontWeight.bold)),
        actions: [
          IconButton(
            icon: Icon(_ocultarValores ? Icons.visibility_off : Icons.visibility),
            onPressed: () => setState(() => _ocultarValores = !_ocultarValores),
          ),
          IconButton(
            icon: _carregandoCotacoes
                ? const SizedBox(width: 20, height: 20, child: CircularProgressIndicator(strokeWidth: 2))
                : const Icon(Icons.refresh),
            onPressed: _atualizarCotacoes,
          ),
        ],
      ),
      body: ListView(
        padding: const EdgeInsets.all(16),
        children: [
          _buildResumoPatrimonio(),
          const SizedBox(height: 20),
          _buildGraficoAlocacao(),
          const SizedBox(height: 24),
          Row(
            mainAxisAlignment: MainAxisAlignment.spaceBetween,
            children: [
              const Text('Meus Ativos', style: TextStyle(fontSize: 18, fontWeight: FontWeight.bold)),
              ElevatedButton.icon(
                style: ElevatedButton.styleFrom(backgroundColor: const Color(0xFF00E676)),
                onPressed: _exibirModalAdicionarAtivo,
                icon: const Icon(Icons.add, color: Colors.black),
                label: const Text('Adicionar', style: TextStyle(color: Colors.black, fontWeight: FontWeight.bold)),
              ),
            ],
          ),
          const SizedBox(height: 12),
          if (widget.carteira.isEmpty)
            const Padding(
              padding: EdgeInsets.symmetric(vertical: 30),
              child: Center(child: Text('Nenhum ativo cadastrado. Clique em Adicionar.', style: TextStyle(color: Colors.white54))),
            )
          else
            ...List.generate(widget.carteira.length, (i) => _buildCardAtivo(widget.carteira[i], i)),
        ],
      ),
    );
  }

  Widget _buildResumoPatrimonio() {
    return Container(
      padding: const EdgeInsets.all(20),
      decoration: BoxDecoration(
        color: const Color(0xFF161B22),
        borderRadius: BorderRadius.circular(16),
        border: Border.all(color: Colors.white10),
      ),
      child: Column(
        crossAxisAlignment: CrossAxisAlignment.start,
        children: [
          const Text('Patrimônio Total', style: TextStyle(color: Colors.white60, fontSize: 14)),
          const SizedBox(height: 8),
          Text(
            _ocultarValores ? 'R\$ ••••••••' : 'R\$ ${patrimonioTotal.toStringAsFixed(2)}',
            style: const TextStyle(fontSize: 30, fontWeight: FontWeight.bold, color: Colors.white),
          ),
          const SizedBox(height: 16),
          Row(
            mainAxisAlignment: MainAxisAlignment.spaceBetween,
            children: [
              Column(
                crossAxisAlignment: CrossAxisAlignment.start,
                children: [
                  const Text('Rentabilidade', style: TextStyle(color: Colors.white54, fontSize: 12)),
                  Text(
                    '${rentabilidadePct >= 0 ? "+" : ""}${rentabilidadePct.toStringAsFixed(2)}%',
                    style: TextStyle(
                      color: rentabilidadePct >= 0 ? const Color(0xFF00E676) : Colors.redAccent,
                      fontWeight: FontWeight.bold,
                      fontSize: 16,
                    ),
                  ),
                ],
              ),
              Column(
                crossAxisAlignment: CrossAxisAlignment.end,
                children: [
                  const Text('Lucro Total', style: TextStyle(color: Colors.white54, fontSize: 12)),
                  Text(
                    _ocultarValores ? 'R\$ •••' : 'R\$ ${lucroTotal.toStringAsFixed(2)}',
                    style: TextStyle(
                      color: lucroTotal >= 0 ? const Color(0xFF00E676) : Colors.redAccent,
                      fontWeight: FontWeight.bold,
                      fontSize: 16,
                    ),
                  ),
                ],
              ),
            ],
          ),
        ],
      ),
    );
  }

  Widget _buildGraficoAlocacao() {
    if (patrimonioTotal == 0) return const SizedBox.shrink();

    return Container(
      height: 200,
      padding: const EdgeInsets.all(16),
      decoration: BoxDecoration(
        color: const Color(0xFF161B22),
        borderRadius: BorderRadius.circular(16),
        border: Border.all(color: Colors.white10),
      ),
      child: Row(
        children: [
          Expanded(
            child: PieChart(
              PieChartData(
                sectionsSpace: 4,
                centerSpaceRadius: 40,
                sections: widget.carteira.map((ativo) {
                  final pct = (ativo.valorTotal / patrimonioTotal) * 100;
                  return PieChartSectionData(
                    value: ativo.valorTotal,
                    title: '${pct.toStringAsFixed(0)}%',
                    radius: 35,
                    titleStyle: const TextStyle(fontSize: 12, fontWeight: FontWeight.bold, color: Colors.white),
                  );
                }).toList(),
              ),
            ),
          ),
          Column(
            mainAxisAlignment: MainAxisAlignment.center,
            crossAxisAlignment: CrossAxisAlignment.start,
            children: widget.carteira.map((a) => Padding(
              padding: const EdgeInsets.symmetric(vertical: 4),
              child: Row(
                children: [
                  Text('${a.ticker}: ', style: const TextStyle(fontWeight: FontWeight.bold)),
                  Text('${((a.valorTotal / patrimonioTotal) * 100).toStringAsFixed(1)}%', style: const TextStyle(color: Colors.white70)),
                ],
              ),
            )).toList(),
          )
        ],
      ),
    );
  }

  Widget _buildCardAtivo(AtivoModel ativo, int index) {
    final pesoPct = patrimonioTotal > 0 ? (ativo.valorTotal / patrimonioTotal) * 100 : 0.0;
    final variacaoPositiva = ativo.variacaoDia >= 0;

    return Dismissible(
      key: Key(ativo.ticker + index.toString()),
      direction: DismissDirection.endToStart,
      background: Container(
        alignment: Alignment.centerRight,
        padding: const EdgeInsets.only(right: 20),
        decoration: BoxDecoration(color: Colors.redAccent, borderRadius: BorderRadius.circular(12)),
        child: const Icon(Icons.delete, color: Colors.white),
      ),
      onDismissed: (_) => _removerAtivo(index),
      child: Card(
        margin: const EdgeInsets.only(bottom: 12),
        shape: RoundedRectangleBorder(borderRadius: BorderRadius.circular(12)),
        child: ListTile(
          contentPadding: const EdgeInsets.symmetric(horizontal: 16, vertical: 8),
          onTap: () {
            Navigator.push(
              context,
              MaterialPageRoute(builder: (context) => DetalhesAtivoScreen(ativo: ativo)),
            );
          },
          leading: CircleAvatar(
            backgroundColor: const Color(0xFF00E676).withOpacity(0.1),
            child: Text(ativo.ticker.substring(0, 2), style: const TextStyle(color: Color(0xFF00E676), fontWeight: FontWeight.bold)),
          ),
          title: Row(
            children: [
              Text(ativo.ticker, style: const TextStyle(fontWeight: FontWeight.bold)),
              const SizedBox(width: 8),
              Container(
                padding: const EdgeInsets.symmetric(horizontal: 6, vertical: 2),
                decoration: BoxDecoration(color: Colors.white10, borderRadius: BorderRadius.circular(4)),
                child: Text(ativo.classe, style: const TextStyle(fontSize: 10, color: Colors.white70)),
              )
            ],
          ),
          subtitle: Text('Peso: ${pesoPct.toStringAsFixed(1)}% | Qtd: ${ativo.quantidade.toInt()}', style: const TextStyle(color: Colors.white54, fontSize: 12)),
          trailing: Column(
            mainAxisAlignment: MainAxisAlignment.center,
            crossAxisAlignment: CrossAxisAlignment.end,
            children: [
              Text(
                _ocultarValores ? 'R\$ •••' : 'R\$ ${ativo.precoAtual.toStringAsFixed(2)}',
                style: const TextStyle(fontWeight: FontWeight.bold, fontSize: 15),
              ),
              Text(
                '${variacaoPositiva ? "+" : ""}${ativo.variacaoDia.toStringAsFixed(2)}%',
                style: TextStyle(
                  color: variacaoPositiva ? const Color(0xFF00E676) : Colors.redAccent,
                  fontWeight: FontWeight.bold,
                  fontSize: 12,
                ),
              ),
            ],
          ),
        ),
      ),
    );
  }
}

class ProventosScreen extends StatelessWidget {
  final List<AtivoModel> carteira;

  const ProventosScreen({super.key, required this.carteira});

  double get totalAcumuladoAno {
    double total = 0;
    for (var a in carteira) {
      total += a.quantidade * (a.dy / 100) * (a.precoAtual > 0 ? a.precoAtual : a.precoMedio);
    }
    return total;
  }

  double get mediaMensal => totalAcumuladoAno / 12;

  @override
  Widget build(BuildContext context) {
    return Scaffold(
      appBar: AppBar(
        backgroundColor: const Color(0xFF161B22),
        elevation: 0,
        title: const Text('Extrato de Proventos', style: TextStyle(fontWeight: FontWeight.bold)),
      ),
      body: ListView(
        padding: const EdgeInsets.all(16),
        children: [
          Container(
            padding: const EdgeInsets.all(20),
            decoration: BoxDecoration(
              color: const Color(0xFF161B22),
              borderRadius: BorderRadius.circular(16),
              border: Border.all(color: Colors.white10),
            ),
            child: Column(
              crossAxisAlignment: CrossAxisAlignment.start,
              children: [
                const Text('Projeção de Proventos (12 Meses)', style: TextStyle(color: Colors.white60, fontSize: 13)),
                const SizedBox(height: 8),
                Text(
                  'R\$ ${totalAcumuladoAno.toStringAsFixed(2)}',
                  style: const TextStyle(fontSize: 28, fontWeight: FontWeight.bold, color: Color(0xFF00E676)),
                ),
                const SizedBox(height: 16),
                Row(
                  mainAxisAlignment: MainAxisAlignment.spaceBetween,
                  children: [
                    Column(
                      crossAxisAlignment: CrossAxisAlignment.start,
                      children: [
                        const Text('Média Mensal Estimada', style: TextStyle(color: Colors.white54, fontSize: 11)),
                        Text('R\$ ${mediaMensal.toStringAsFixed(2)}/mês', style: const TextStyle(fontWeight: FontWeight.bold, color: Colors.white)),
                      ],
                    ),
                    const Column(
                      crossAxisAlignment: CrossAxisAlignment.end,
                      children: [
                        Text('Yield on Cost (YOC)', style: TextStyle(color: Colors.white54, fontSize: 11)),
                        Text('10.4%', style: TextStyle(fontWeight: FontWeight.bold, color: Color(0xFF00E676))),
                      ],
                    )
                  ],
                )
              ],
            ),
          ),
          const SizedBox(height: 24),
          const Text('Últimos Lançamentos', style: TextStyle(fontSize: 16, fontWeight: FontWeight.bold)),
          const SizedBox(height: 12),
          if (carteira.isEmpty)
            const Center(child: Text('Cadastre ativos para visualizar os proventos.', style: TextStyle(color: Colors.white54)))
          else
            ...carteira.map((ativo) {
              final valorEstimadoUnitario = (ativo.dy / 100 * ativo.precoAtual) / 4;
              final valorTotalProvento = valorEstimadoUnitario * ativo.quantidade;
              return Container(
                margin: const EdgeInsets.only(bottom: 10),
                padding: const EdgeInsets.all(12),
                decoration: BoxDecoration(
                  color: const Color(0xFF161B22),
                  borderRadius: BorderRadius.circular(12),
                  border: Border.all(color: Colors.white10),
                ),
                child: Row(
                  mainAxisAlignment: MainAxisAlignment.spaceBetween,
                  children: [
                    Row(
                      children: [
                        CircleAvatar(
                          backgroundColor: const Color(0xFF00E676).withOpacity(0.1),
                          child: Text(ativo.ticker.substring(0, 2), style: const TextStyle(color: Color(0xFF00E676), fontWeight: FontWeight.bold, fontSize: 12)),
                        ),
                        const SizedBox(width: 12),
                        Column(
                          crossAxisAlignment: CrossAxisAlignment.start,
                          children: [
                            Text(ativo.ticker, style: const TextStyle(fontWeight: FontWeight.bold)),
                            Text('${ativo.classe == "FIIs" ? "Rendimento" : "Dividendo"} • ${ativo.quantidade.toInt()} cotas', style: const TextStyle(color: Colors.white54, fontSize: 11)),
                          ],
                        )
                      ],
                    ),
                    Text('R\$ ${valorTotalProvento.toStringAsFixed(2)}', style: const TextStyle(fontWeight: FontWeight.bold, color: Color(0xFF00E676), fontSize: 15)),
                  ],
                ),
              );
            }).toList(),
        ],
      ),
    );
  }
}

class DetalhesAtivoScreen extends StatelessWidget {
  final AtivoModel ativo;

  const DetalhesAtivoScreen({super.key, required this.ativo});

  @override
  Widget build(BuildContext context) {
    return Scaffold(
      appBar: AppBar(
        backgroundColor: const Color(0xFF161B22),
        elevation: 0,
        title: Text('${ativo.ticker} - Ficha Técnica', style: const TextStyle(fontWeight: FontWeight.bold)),
      ),
      body: ListView(
        padding: const EdgeInsets.all(16),
        children: [
          Container(
            padding: const EdgeInsets.all(20),
            decoration: BoxDecoration(
              color: const Color(0xFF161B22),
              borderRadius: BorderRadius.circular(16),
              border: Border.all(color: Colors.white10),
            ),
            child: Column(
              crossAxisAlignment: CrossAxisAlignment.start,
              children: [
                Text(ativo.nome, style: const TextStyle(color: Colors.white54, fontSize: 14)),
                const SizedBox(height: 4),
                Text('R\$ ${ativo.precoAtual.toStringAsFixed(2)}', style: const TextStyle(fontSize: 28, fontWeight: FontWeight.bold, color: Colors.white)),
              ],
            ),
          ),
          const SizedBox(height: 24),
          const Text('Indicadores Fundamentalistas', style: TextStyle(fontSize: 16, fontWeight: FontWeight.bold)),
          const SizedBox(height: 12),
          GridView.count(
            crossAxisCount: 2,
            shrinkWrap: true,
            physics: const NeverScrollableScrollPhysics(),
            childAspectRatio: 2.1,
            crossAxisSpacing: 12,
            mainAxisSpacing: 12,
            children: [
              _buildCardIndicador('P/L (Preço / Lucro)', ativo.pl > 0 ? ativo.pl.toStringAsFixed(2) : '-'),
              _buildCardIndicador('Dividend Yield (DY)', '${ativo.dy.toStringAsFixed(2)}%'),
              _buildCardIndicador('P/VP (Preço / V. Patr.)', ativo.pvp.toStringAsFixed(2)),
              _buildCardIndicador('ROE (Retorno s/ Patr.)', '${ativo.roe.toStringAsFixed(2)}%'),
            ],
          ),
        ],
      ),
    );
  }

  Widget _buildCardIndicador(String titulo, String valor) {
    return Container(
      padding: const EdgeInsets.all(12),
      decoration: BoxDecoration(
        color: const Color(0xFF161B22),
        borderRadius: BorderRadius.circular(12),
        border: Border.all(color: Colors.white10),
      ),
      child: Column(
        crossAxisAlignment: CrossAxisAlignment.start,
        mainAxisAlignment: MainAxisAlignment.center,
        children: [
          Text(titulo, style: const TextStyle(color: Colors.white54, fontSize: 11)),
          const SizedBox(height: 6),
          Text(valor, style: const TextStyle(fontSize: 16, fontWeight: FontWeight.bold, color: Colors.white)),
        ],
      ),
    );
  }
}
```
